## Bagian A — Linear Regression

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, confusion_matrix, classification_report
from sklearn.datasets import load_breast_cancer

rng = np.random.default_rng(42)
X = rng.uniform(20, 120, size=(200,1))
y = 50 + 1.8*X[:,0] + rng.normal(0, 18, size=200)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.25, random_state=42)
lin = LinearRegression().fit(X_train, y_train)
pred = lin.predict(X_test)
print('intercept=', lin.intercept_, 'coef=', lin.coef_[0])
print('MAE=', mean_absolute_error(y_test,pred))
print('RMSE=', mean_squared_error(y_test,pred)**0.5)
```

**Interpretasi Koefisien:**

Nilai koefisien (sekitar 1.8) menunjukkan bahwa untuk setiap kenaikan 1 satuan pada variabel $X$, nilai $y$ diprediksi naik sebesar 1.8 satuan. Angka ini hanya menunjukkan pola korelasi matematis antar variabel, dan bukan berarti $X$ adalah penyebab pasti naiknya $y$ (bukan hubungan kausalitas).

**3 Residual Terbesar:**

Berdasarkan *output* tabel di atas, terdapat 3 titik data dengan nilai residual (selisih absolut antara nilai prediksi dan aktual) yang paling besar. Ini menjadi *evidence* bahwa garis regresi linier tidak memotong semua titik data secara sempurna dan model masih menyisakan *error* pada beberapa titik ekstrem.

Interpretasikan koefisien tanpa mengklaim kausalitas. Cari 3 residual terbesar.

```python
resid = y_test - pred
idx = np.argsort(np.abs(resid))[-3:]
pd.DataFrame({'aktual':y_test[idx], 'prediksi':pred[idx], 'residual':resid[idx]})
```

**Interpretasi Koefisien:**
Nilai koefisien yang dihasilkan (sekitar 1.8) menunjukkan pola korelasi: setiap kenaikan 1 satuan pada variabel $X$, nilai $y$ diprediksi akan naik sebesar 1.8 satuan. Namun, angka ini hanya merepresentasikan hubungan matematis pada data, bukan kausalitas. Artinya, kita tidak bisa mengklaim bahwa $X$ adalah penyebab pasti dari perubahan $y$.

**3 Residual Terbesar:**
Berdasarkan *output* tabel kode, terdapat 3 titik data dengan nilai residual (selisih absolut antara nilai prediksi dan aktual) yang paling besar/meleset. Ini menjadi *evidence* bahwa garis regresi linier tidak memotong semua titik data secara sempurna dan model masih menyisakan *error* pada beberapa titik ekstrem.

## Bagian B — Logistic Regression

```python
data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(
    data.data, data.target, test_size=.25, random_state=42, stratify=data.target
)
log = LogisticRegression(max_iter=5000).fit(X_train, y_train)
proba = log.predict_proba(X_test)[:,1]

for t in [0.3, 0.5, 0.7]:
    pred_t = (proba >= t).astype(int)
    print('\nthreshold', t)
    print(confusion_matrix(y_test,pred_t))
    print(classification_report(y_test,pred_t, digits=3))
```

Bandingkan threshold 0.3, 0.5, 0.7. Apa yang berubah pada false positive dan false negative?

Mengubah nilai *threshold* menciptakan *trade-off* antara *precision* dan *recall*:

* **Threshold 0.3 (Diturunkan):** Model menjadi lebih mudah dan longgar dalam memprediksi kelas positif. Akibatnya, jumlah *False Positive* (salah menebak positif) **meningkat**, sedangkan *False Negative* (gagal mendeteksi positif) **menurun**.

* **Threshold 0.5:** Merupakan titik batas keseimbangan standar (*default*).

* **Threshold 0.7 (Dinaikkan):** Model menjadi sangat ketat sebelum memprediksi kelas positif. Akibatnya, jumlah *False Positive* akan **menurun**, sedangkan jumlah *False Negative* akan **meningkat**.
